```python
import string

lowercase = list(string.ascii_lowercase)
uppercase = list(string.ascii_uppercase)
numbers = list(string.digits)
symbols = list(string.punctuation)


Characterbucket=lowercase+uppercase+numbers+symbols

import random

password=[]
for i in range (1,13):
 i=random.choice(Characterbucket)
 password.append(i)
print(password)

```
