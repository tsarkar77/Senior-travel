```python
import random
number=random.randint(1,10)

def guessgame():
 tries = 0
 Guess=int(input("Guess between 1-10:"))

 while Guess!=number:
  Guess=int(input("Try again:"))
  tries+=1

 if int(Guess)==number:
    print(f"CONGRATS you got it in {tries} tries!")

guessgame()








```
